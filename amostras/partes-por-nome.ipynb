# Contagem de partes por nome

```python
from collections import Counter

total = 0
dois_nomes = 0
part_count = Counter()

with open('nomes-reais.txt', encoding='ascii') as fp:
    for line in fp:
        total += 1
        part_count[len(line.strip().split())] += 1

        
print(part_count.most_common())
```

```python
total = float(total)
for parts, count in part_count.most_common():
    print(f"{parts}\t{count/total*100:4.1f}%")
```
